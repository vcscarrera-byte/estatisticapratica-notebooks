## Gráficos de barras e setores

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/cibelerusso/Datasets/main/Music_DB.csv', decimal=',')
```

```python
# Tabela de frequências absolutas

tab = pd.crosstab(index=df['Compasso'], columns='frequência')

tab
```

```python
# Define o tamanho da figura e a resolução
plt.figure(figsize=(20, 12), dpi=100)  # Aumenta a largura e a altura da figura

tab.plot.bar(
    color='skyblue',           # Define a cor das barras
    edgecolor='black',         # Adiciona cor às bordas das barras
    legend=False,              # Remove a legenda se houver apenas uma série
)
```

```python
import matplotlib.pyplot as plt

# Supondo que 'tab' seja uma tabela ou DataFrame que já foi definido
ax = tab.plot.bar(
    color='skyblue',
    legend=False,
)

plt.savefig('grafico_barras_compasso.png', dpi=300)  # Salva com 300 dpi


# Ajusta o tamanho do gráfico depois de criá-lo
fig = plt.gcf()  # Obtém a figura atual
fig.set_size_inches(12, 6)  # Define a largura e altura desejadas

# Adiciona um título ao gráfico
plt.title('Gráfico de barras de Compasso', fontsize=16)

# Adiciona rótulo para o eixo y
plt.ylabel('Frequência', fontsize=12)

# Adiciona rótulo para o eixo x
plt.xlabel('Categorias', fontsize=12)

# Rotaciona os rótulos do eixo x para melhor legibilidade
plt.xticks(rotation=45, ha='right', fontsize=12)

# Adiciona rótulos nas barras indicando os valores
for container in ax.containers:
    ax.bar_label(container, fmt='%.0f', label_type='edge', fontsize=12)

# Ajusta o limite do eixo y
plt.ylim(0, tab.max().max() * 1.2)

# Ajusta layout para evitar sobreposição
plt.tight_layout()

# Exibe o gráfico
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Configuração da figura com subplots
fig, axes = plt.subplots(2, 2, figsize=(14, 8), gridspec_kw={'height_ratios': [1, 3]})

# Boxplots para 'Energia' e 'Orgânico'
sns.boxplot(x=df['Energia'], ax=axes[0, 0], color='red')
sns.boxplot(x=df['Orgânico'], ax=axes[0, 1], color='red')

# Histogramas com densidade alisada para 'Energia' e 'Orgânico'
sns.histplot(df['Energia'], kde=True, ax=axes[1, 0], color='red')
sns.histplot(df['Orgânico'], kde=True, ax=axes[1, 1], color='red')

# Customização dos eixos
axes[0, 0].set(xlabel='')  # Remove o label do eixo x no boxplot de 'Energia'
axes[0, 1].set(xlabel='')  # Remove o label do eixo x no boxplot de 'Orgânico'
axes[1, 0].set_ylabel('Frequência')  # Adiciona label ao eixo y do histograma de 'Energia'
axes[1, 1].set_ylabel('Frequência')  # Adiciona label ao eixo y do histograma de 'Orgânico'

# Títulos dos gráficos
axes[0, 0].set_title('Boxplot de Energia')
axes[0, 1].set_title('Boxplot de Orgânico')
axes[1, 0].set_title('Histograma de Energia')
axes[1, 1].set_title('Histograma de Orgânico')

# Ajuste da disposição dos gráficos
plt.tight_layout()

# Exibição do gráfico
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Configuração da figura para acomodar todos os boxplots
plt.figure(figsize=(12, 8))

Quanti = df[['Energia','Ao_vivo','Falado','Orgânico','Instrumental','Dançabilidade',
             'Força','Positividade']]

# Criação dos boxplots para todas as colunas em 'Quanti'
sns.boxplot(data=Quanti, palette='Reds')

# Customização do eixo x
plt.xticks(rotation=45, ha='right')  # Rotaciona os rótulos das colunas para melhor legibilidade
plt.ylabel('Valores')

# Título do gráfico
plt.title('Boxplots para Variáveis Quantitativas')

# Exibição do gráfico
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Configuração da figura para acomodar todos os boxplots
plt.figure(figsize=(12, 8))

# Criação dos boxplots para todas as colunas em 'Quanti'
sns.boxplot(data=Quanti, palette='Reds')

# Customização do eixo x
plt.xticks(rotation=45, ha='right')  # Rotaciona os rótulos das colunas para melhor legibilidade
plt.ylabel('Valores')

# Título do gráfico
plt.title('Boxplots para Variáveis Quantitativas')

# Exibição do gráfico
plt.show()
```

```python
# Define o número de colunas e linhas para os subplots
num_columns = 5
num_rows = len(Quanti.columns) // num_columns + (len(Quanti.columns) % num_columns > 0)

# Configura o tamanho da figura
fig, axes = plt.subplots(num_rows, num_columns, figsize=(15, num_rows * 3))

# Plota cada histograma
for i, column in enumerate(Quanti.columns):
    row = i // num_columns
    col = i % num_columns
    axes[row, col].hist(Quanti[column], bins=20)
    axes[row, col].set_title(f'Histograma de {column}')
    axes[row, col].set_xlabel(column)
    axes[row, col].set_ylabel('Frequência')

# Ajusta o layout para não sobrepor os subplots
plt.tight_layout()
plt.show()
```

```python

# Define o número de colunas e linhas para os subplots
num_columns = 5
num_rows = len(Quanti.columns) // num_columns + (len(Quanti.columns) % num_columns > 0)

# Configura o tamanho da figura
fig, axes = plt.subplots(num_rows, num_columns, figsize=(15, num_rows * 3))

# Plota cada densidade alisada
for i, column in enumerate(Quanti.columns):
    row = i // num_columns
    col = i % num_columns
    sns.kdeplot(Quanti[column], ax=axes[row, col], fill=True, color='blue', alpha=0.6)
    axes[row, col].set_title(f'Densidade Alisada de {column}')
    axes[row, col].set_xlabel(column)
    axes[row, col].set_ylabel('Densidade')

# Ajusta o layout para não sobrepor os subplots
plt.tight_layout()
plt.show()
```
