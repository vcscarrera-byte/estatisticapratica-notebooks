## Tabelas de frequências

```python
# Tabela de frequências absolutas

tab = pd.crosstab(index=df['Compasso'], columns='frequência')

tab
```

```python
# Tabela de frequências relativas

tab = pd.crosstab(index=df['Compasso'], columns='frequência', normalize='columns')

tab
```

```python
# Tabela de frequências relativas

tab = pd.crosstab(index=df['Compasso'], columns='frequência', margins='columns', margins_name='Total')

tab
```

```python
# Tabela de frequências relativas

tab = pd.crosstab(index=df['Playlist_name'], columns='frequência')

tab
```

```python
# Tabela de frequências relativas

tab = pd.crosstab(index=df['Playlist_name'], columns='frequência', normalize='columns')

tab
```
