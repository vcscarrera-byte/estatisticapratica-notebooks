### Gráficos de mosaico

```python
import pandas as pd
from statsmodels.graphics.mosaicplot import mosaic
import matplotlib.pyplot as plt


# Cria a figura e os eixos
fig, ax = plt.subplots(figsize=(15, 10))

# Cria o gráfico de mosaico
mosaic(df, ['Tom', 'Modo'], title='Gráfico de Mosaico: Tom por Modo',
       properties=dict(boxdict={'facecolor': 'lightblue', 'edgecolor': 'black', 'linewidth': 2}),
       labelizer=lambda k: f'{k[0]}\n{k[1]}',
       gap=0.01, ax=ax)

# Ajusta os rótulos e título
ax.set_title('Gráfico de Mosaico: Tom por Modo', fontsize=16)
ax.set_xlabel('Tom', fontsize=14)
ax.set_ylabel('Modo', fontsize=14)

# Ajusta o layout e exibe o gráfico
plt.show()
```

```python
import plotly.graph_objects as go
from sklearn.preprocessing import StandardScaler
import pandas as pd

# Seleciona as colunas quantitativas
Quanti = df[['Energia', 'Ao_vivo', 'Tempo', 'Falado', 'Orgânico', 'Instrumental', 'Dançabilidade', 'Duração', 'Força', 'Positividade']]

# Padroniza as variáveis
scaler = StandardScaler()
Quanti_scaled = scaler.fit_transform(Quanti)

# Converte para DataFrame
Quanti_scaled = pd.DataFrame(Quanti_scaled, columns=Quanti.columns)

# Escolhe a observação (por exemplo, a primeira linha)
observation_index = 0  # Índice da observação desejada
observation = Quanti_scaled.iloc[observation_index]

# Cria o gráfico de radar
fig = go.Figure()

fig.add_trace(go.Scatterpolar(
    r=observation.values,
    theta=Quanti.columns,
    fill='toself',
    name=f'Observação {observation_index + 1}'
))

# Ajusta o layout do gráfico
fig.update_layout(
    polar=dict(
        radialaxis=dict(visible=True)
    ),
    showlegend=True
)

# Exibe o gráfico
fig.show()
```

```python
import plotly.graph_objects as go
from sklearn.preprocessing import StandardScaler
import pandas as pd
import plotly.subplots as sp


# Seleciona as colunas quantitativas
Quanti = df[['Energia', 'Ao_vivo', 'Tempo', 'Falado', 'Orgânico', 'Instrumental', 'Dançabilidade', 'Duração', 'Força', 'Positividade']]

# Padroniza as variáveis
scaler = StandardScaler()
Quanti_scaled = scaler.fit_transform(Quanti)

# Converte para DataFrame
Quanti_scaled = pd.DataFrame(Quanti_scaled, columns=Quanti.columns)

# Fixa a semente para reprodutibilidade
np.random.seed(42)

# Seleciona 6 observações aleatórias
random_indices = np.random.choice(Quanti_scaled.index, size=6, replace=False)

# Cria subplots com layout 2x3
fig = sp.make_subplots(rows=2, cols=3, subplot_titles=[f'Observação {index}' for index in random_indices],
                       specs=[[{'type': 'polar'}, {'type': 'polar'}, {'type': 'polar'}],
                              [{'type': 'polar'}, {'type': 'polar'}, {'type': 'polar'}]])

# Adiciona os gráficos de radar para cada uma das observações selecionadas
for i, observation_index in enumerate(random_indices):
    observation = Quanti_scaled.iloc[observation_index]

    fig.add_trace(go.Scatterpolar(
        r=observation.values,
        theta=Quanti.columns,
        fill='toself',
        name=f'Observação {observation_index}'
    ), row=(i//3)+1, col=(i%3)+1)

# Ajusta o layout do gráfico
fig.update_layout(height=1000, width=1500, title_text="Gráficos de Radar para 6 observações aleatoriamente selecionadas")

# Exibe o gráfico
fig.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Supondo que você já tenha um DataFrame chamado 'Quanti'
# Calcular a matriz de correlações de Pearson
correlation_matrix = Quanti.corr(method='pearson')

# Criar o mapa de calor
plt.figure(figsize=(12, 8))  # Definindo o tamanho da figura
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap='coolwarm', square=True, cbar=True)

# Exibir o gráfico
plt.title('Matriz de Correlações de Pearson')
plt.show()
```
